```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread(r'C:\Users\arpit\Desktop\Image_Fundamentals\Udemy_Pierian_Data_OpenCV\Computer-Vision-with-Python\DATA\internal_external.png',0)
```

```python
plt.imshow(img,cmap='gray')
```

```python
contours, heirarchy = cv2.findContours(img,cv2.RETR_CCOMP, cv2.CHAIN_APPROX_SIMPLE)
```

```python
type(contours)
```

```python
len(contours)
```

```python
type(heirarchy)
```

```python
external_contours = np.zeros(img.shape)
```

```python
external_contours.shape
```

```python
for i in range(len(contours)):

  # External
  if heirarchy[0][i][3] == -1:
    cv2.drawContours(external_contours,contours,i,255,-1)
```

```python
plt.imshow(external_contours,cmap='gray')
```

```python
internal_contours = np.zeros(img.shape)
```

```python
for i in range(len(contours)):

  # Internal
  if heirarchy[0][i][3] != -1:
    cv2.drawContours(internal_contours,contours,i,255,-1)
```

```python
plt.imshow(internal_contours,cmap='gray')
```

```python
All = external_contours - internal_contours
```

```python
plt.imshow(All,cmap='gray')
```
