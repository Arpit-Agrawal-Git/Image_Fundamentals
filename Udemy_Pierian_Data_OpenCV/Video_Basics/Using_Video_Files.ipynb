```python
import cv2
import time
```

```python
cap = cv2.VideoCapture('myvideo.mp4')
```

```python
if cap.isOpened() == False:
  print("Error! FILE NOT FOUND OR WRONG CODEC USED")

while cap.isOpened():

  ret, frame = cap.read()

  if ret == True :

    # Writer 20  FPS
    time.sleep(1/20)
    cv2.imshow('frame', frame)

    if cv2.waitKey(1) & 0xFF == 27:
      break

  else:
    break

cap.release()
cv2.destroyAllWindows()
```
