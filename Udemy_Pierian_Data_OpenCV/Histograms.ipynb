```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
dark_horse = cv2.imread(r'Computer-Vision-with-Python\DATA\horse.jpg')
show_horse = cv2.cvtColor(dark_horse, cv2.COLOR_BGR2RGB)

rainbow = cv2.imread(r'Computer-Vision-with-Python\DATA\rainbow.jpg')
show_rainbow = cv2.cvtColor(rainbow, cv2.COLOR_BGR2RGB)

blue_bricks = cv2.imread(r'Computer-Vision-with-Python\DATA\bricks.jpg')
show_bricks = cv2.cvtColor(blue_bricks, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(show_horse)
```

```python
hist_values = cv2.calcHist([blue_bricks],channels=[0],mask=None,histSize=[256], ranges=[0,256] )
```

```python
plt.plot(hist_values)
```

```python
horse_hist = cv2.calcHist([dark_horse],channels=[2],mask=None,histSize=[256], ranges=[0,256] )
```

```python
plt.plot(horse_hist)
```

```python
img = rainbow
```

```python
color = ('b','g','r')

for i,col in enumerate(color):
  histr = cv2.calcHist([img],[i],None,[256],[0,256])
  plt.plot(histr, color=col)
  plt.xlim([0,256])

plt.title('Histogram for Blue Bricks')
```
